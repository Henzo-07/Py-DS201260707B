```python
# To establish a connection between Python and MYSQL
try:
    db = pymysql.connect(
        user = "root",
        port = 3306,
        host = "localhost",
        password = PASSWORD
    )
    cursor = db.cursor()
    if cursor:
        print("Connection  Successful..")
except Exception as e:
    print(f"Error message: {e}")
```

```python
# To show all databases using Python.
cursor.execute("SHOW databases")
for i in cursor:
    print(i)
```

```python
# To create a database in Python. 
cursor.execute("CREATE DATABASE AIRLINE")
cursor.execute("SHOW databases")
for i in cursor:
    print(i)
```

```python
cursor.execute("USE airline")
try:
    cursor.execute("""
    CREATE TABLE FLIGHT(
    Flight_id INT PRIMARY KEY AUTO_INCREMENT, 
    DESTINATION VARCHAR(25) NOT NULL, 
    DEPARTURE VARCHAR(25) NOT NULL,
    FLIGHT_MADE DATETIME DEFAULT CURRENT_TIMESTAMP
    )""")
    if cursor:
        print("Table created..")
except Exception as e:
    print(f"Error Message:{e}")
```

```python
try:
    cursor.execute("""
    CREATE TABLE CUSTOMERS(
    CUSTOMER_id VARCHAR(10) PRIMARY KEY, 
    FULL_NAME VARCHAR(25) UNIQUE NOT NULL, 
    AGE INT NOT NULL,
    EMAIL TEXT NOT NULL,
    DAY_REGISTRATION DATETIME DEFAULT CURRENT_TIMESTAMP
    )""")
    if cursor:
        print("Table created..")
except Exception as e:
    print(f"Error Message:{e}")
```

```python
try:
    cursor.execute("""
    CREATE TABLE BOOKINGS(
    BOOKING_id VARCHAR(10) PRIMARY KEY, 
    Flight_id INT NOT NULL, 
    Customer_ID VARCHAR(10) NOT NULL,
    TRAVEL_DATE DATE NOT NULL,
    FOREIGN KEY(Flight_id) REFERENCES FLIGHT (Flight_id),
    FOREIGN KEY(CUSTOMER_id) REFERENCES CUSTOMERS (Customer_id)
    )""")
    if cursor:
        print("Table created..")
except Exception as e:
    print(f"Error Message:{e}")
```
